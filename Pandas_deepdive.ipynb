#### handles division by 0 by try - except

```python
num1 = input("Enter numeratoer:")
num2= input("Enter denominator")

try:
    result = int(num1)/int(num2)
    print(result)
except ZeroDivisionError:
    print("Denominator cannot be zero")

except ValueError:
    print("Enter num only")

finally:
    print("Completed!")

```

### Count how many vowels in a given string

```python
userinput= input("Enter characters:")
vowels = "aeiou"
count =0

for i in userinput.lower():
    if i in vowels:
        count=count+1
    print(i)

print ("total count = ",count)
```

### Reverse a list without using built-in function

```python
Oglist= [4,3,7,8,9,21]
Oglist
```

```python
Rlist =[]

a = len(Oglist)-1

while a>=0:
    Rlist.append(Oglist(a))
    a-=1

Rlist

    
```
