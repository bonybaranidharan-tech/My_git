#### 1.List
emptylist [] - square bracket is used for list

* 1 - Dimensional
* Hetrogenous - collection of multiple datatypes like string , float (5.5.7.5).
* Duplicate values
* Indexing & slicing is posible - Index no. always starts from 0 from front - from backwards -1
* Mutable - values once assinged to the variable can be replaced

Use [] - to get a piece of information
  

```python
to_buy = ['apple','orange','grapes','pomo',4.3,3,5.4]
```

```python
type(to_buy)
```

```python
to_buy[3] #indexing - from left starts from 0
```

```python
to_buy[-5] #indexing - from right starts from -1
```

```python
to_buy[:4] #prints 0 to 3 in position
```

```python
to_buy[3:6] #prints 3,4,5 positions
```

```python
to_buy[-5:-2]
```

```python
to_buy
```

```python
to_buy[2]= 'watermelon'
```
